### My next steps:
- Get rid of the excess rows so that i am only keeping the 23 seasons that Lebron has played
- Fixing data types

```python
#.str.isdigit() is True when the text is only digits, like 19, and False for 23 Yrs.
#.fillna(False) treats blank cells as False.
#how it works, it keeps going until there is no age, so obviously the age stops at lebron's latest season.
#23 columns
dfcopy = dfcopy[dfcopy['Age'].str.isdigit().fillna(False)]
dfcopy
```

```python
dfcopy.info()
#now running the .info, you can now see that there are 23 rows which is the amount of seasons Lebron has played
#BUT every column is still a string data type
```

```python
# from the data above, I want the columns SEASON, TEAM, LG, POS, AWARDS to stay as text
text_cols= ['Season', 'Team', 'Lg', 'Pos', 'Awards']
text_cols
```

```python
#every other column not those should be a number/integer
num_cols = [col for col in dfcopy.columns if col not in text_cols]
num_cols
```

```python
#convert them all at once
#.apply(pd.to_numeric) converts each col into numeric datatype
dfcopy[num_cols] = dfcopy[num_cols].apply(pd.to_numeric)
dfcopy.info()
```

## Next Steps:
### - Okay, now that I have tackled the data types, as well as keeping the only 23 rows of Lebron's 23 years of basketball - I am ready to now bring this to SQL
- But, before that, the columns FG%, 3p, 2p% have the % which will be problematic in SQL, since % is usually a wildcard in SQL. So, I want every column to be lowercase, with no % and also not starting with a digit.
- Goal is that % becomes _pct, 3p becomes three_p, 2p becomes two_p

```python
#Look at the current names of the csv 
dfcopy.columns

#i can see that there are 9 columns that need to be fixed in terms of % and having number digits
#and also every column needs to be lower case
```

```python
dfcopy.columns = (
    dfcopy.columns
    .str.lower() #FG% to fg%
    .str.replace('%','_pct') #fg% to fg_pct
    .str.replace('3p','three_p') #3p to three_p
    .str.replace('2p','two_p') #2p to two_p
)
dfcopy.columns
```
